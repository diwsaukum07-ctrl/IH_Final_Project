```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
matches_cleaned = pd.read_csv("matches_cleaned.csv")
matches_cleaned
```

```python
players_cleaned = pd.read_csv("players_cleaned.csv")
```

```python
players_cleaned = players_cleaned[["match_id", "player_slot", "hero_id"]]
```

```python
players_cleaned
```

```python
#players_cleaned.to_csv("players_cleaned.csv", index=False)
```

```python
players_cleaned["player_slot"].value_counts()
```

```python
players_aggregated = players_cleaned.pivot(
    index="match_id",
    columns="player_slot",
    values="hero_id"
).reset_index()
```

```python
players_aggregated
```

```python
players_aggregated = players_aggregated.rename(columns={
    0: "Radiant_1",
    1: "Radiant_2",
    2: "Radiant_3",
    3: "Radiant_4",
    4: "Radiant_5",
    128: "Dire_1",
    129: "Dire_2",
    130: "Dire_3",
    131: "Dire_4",
    132: "Dire_5"
})
```

```python
players_aggregated
```

```python
#players_aggregated.to_csv("players_aggregated.csv", index=False)
```

```python
heroes_cleaned = pd.read_csv("heroes_cleaned.csv")
heroes_cleaned
```

```python
final_dataset = players_aggregated.merge(
    matches_cleaned[["match_id", "radiant_win"]],
    on="match_id",
    how="inner"
)
```
