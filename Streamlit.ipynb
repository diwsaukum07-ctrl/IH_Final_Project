```python
import streamlit as st
import pandas as pd
import numpy as np
import joblib
```

```python
st.title("Dota 2 Match Outcome Predictor")

st.write("Select 5 Radiant heroes and 5 Dire heroes.")

# Load hero names
heroes = pd.read_csv("heroes_cleaned.csv")

hero_names = heroes["localized_name"].tolist()

st.subheader("Radiant")

radiant_1 = st.selectbox("Radiant Hero 1", hero_names)
radiant_2 = st.selectbox("Radiant Hero 2", hero_names)
radiant_3 = st.selectbox("Radiant Hero 3", hero_names)
radiant_4 = st.selectbox("Radiant Hero 4", hero_names)
radiant_5 = st.selectbox("Radiant Hero 5", hero_names)

st.subheader("Dire")

dire_1 = st.selectbox("Dire Hero 1", hero_names)
dire_2 = st.selectbox("Dire Hero 2", hero_names)
dire_3 = st.selectbox("Dire Hero 3", hero_names)
dire_4 = st.selectbox("Dire Hero 4", hero_names)
dire_5 = st.selectbox("Dire Hero 5", hero_names)
```

```python
model = joblib.load("dota_model.pkl")
scaler = joblib.load("scaler.pkl")
feature_columns = joblib.load("feature_columns.pkl")

radiant_team = [radiant_1,radiant_2,radiant_3,radiant_4,radiant_5]

dire_team = [dire_1,dire_2,dire_3,dire_4,dire_5]
```

```python
if st.button("Predict Winner"):
    st.write("Radiant:", radiant_team)
    st.write("Dire:", dire_team)
```
